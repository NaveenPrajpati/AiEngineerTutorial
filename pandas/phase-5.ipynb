# Phase 5 — Apply, Lambda & Visualization

Topics:
- `apply` — run a function on each row or column
- `map` — transform values in a Series
- `lambda` — inline functions
- `pd.cut` / `pd.qcut` — binning continuous data
- Plotting with pandas (built on matplotlib)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    'Name':       ['ram', 'shyam', 'karan', 'raju', 'priya', 'meena', 'arjun', 'sita'],
    'Department': ['IT',  'HR',    'IT',    'HR',   'IT',    'Finance','Finance','HR'],
    'salary':     [50000, 40000,   60000,   45000,  55000,   70000,   65000,   42000],
    'experience': [3,     5,       7,       2,      4,       9,       8,       3],
    'rating':     [4.2,   3.8,     4.5,     3.5,    4.0,     4.8,     4.6,     3.7]
})
print(df)
```

## 1. apply — Run a Function on Each Row or Column

`apply(func, axis=0)` applies `func` to each **column** (default).  
`apply(func, axis=1)` applies `func` to each **row**.

```python
# Apply to a single column — convert salary to annual
df['annual_salary'] = df['salary'].apply(lambda x: x * 12)
print(df[['Name', 'salary', 'annual_salary']])
```

```python
# Apply a named function — categorize experience level
def experience_level(years):
    if years <= 3:
        return 'Junior'
    elif years <= 6:
        return 'Mid'
    else:
        return 'Senior'

df['level'] = df['experience'].apply(experience_level)
print(df[['Name', 'experience', 'level']])
```

```python
# apply row-wise (axis=1) — combine values from multiple columns
df['summary'] = df.apply(
    lambda row: f"{row['Name'].title()} ({row['Department']}) — ₹{row['salary']:,}",
    axis=1
)
print(df[['summary']])
```

## 2. map — Transform Values in a Series

`map` works on a **Series** only. Pass a dictionary or a function.  
Great for encoding categorical values or replacing labels.

```python
# Map department names to codes
dept_code = {'IT': 1, 'HR': 2, 'Finance': 3}
df['dept_code'] = df['Department'].map(dept_code)
print(df[['Name', 'Department', 'dept_code']])

# Map with a function — uppercase department
df['dept_upper'] = df['Department'].map(str.upper)
print(df[['Department', 'dept_upper']])
```

## 3. Binning — pd.cut and pd.qcut

Convert continuous numbers into categories (bins).

| Function | How bins are defined |
|---|---|
| `pd.cut` | Fixed-width bins you define |
| `pd.qcut` | Equal-frequency bins (same number of records per bin) |

```python
# pd.cut — fixed salary bands
df['salary_band'] = pd.cut(
    df['salary'],
    bins=[0, 45000, 60000, 100000],
    labels=['Low', 'Mid', 'High']
)
print(df[['Name', 'salary', 'salary_band']])

# pd.qcut — split into 3 equal-sized groups
df['salary_tier'] = pd.qcut(df['salary'], q=3, labels=['Bottom', 'Middle', 'Top'])
print("\n", df[['Name', 'salary', 'salary_tier']])
```

## 4. Visualization

Pandas wraps matplotlib to let you plot directly from a DataFrame.  
For quick exploratory charts, this is much faster than raw matplotlib.

```python
# Bar chart — average salary per department
avg_salary = df.groupby('Department')['salary'].mean()
avg_salary.plot(kind='bar', color='steelblue', edgecolor='white', figsize=(8, 4))
plt.title('Average Salary by Department')
plt.ylabel('Salary (₹)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Histogram — salary distribution
df['salary'].plot(kind='hist', bins=6, color='coral', edgecolor='white', figsize=(8, 4))
plt.title('Salary Distribution')
plt.xlabel('Salary (₹)')
plt.tight_layout()
plt.show()
```

```python
# Scatter plot — experience vs salary
df.plot(kind='scatter', x='experience', y='salary', color='mediumseagreen',
        figsize=(8, 4), s=100)
plt.title('Experience vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary (₹)')
plt.tight_layout()
plt.show()
```

```python
# Pie chart — department headcount
dept_count = df['Department'].value_counts()
dept_count.plot(kind='pie', autopct='%1.0f%%', figsize=(6, 6), startangle=90)
plt.title('Department Headcount')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Box plot — salary spread per department
df.boxplot(column='salary', by='Department', figsize=(8, 5), grid=False)
plt.title('Salary Spread by Department')
plt.suptitle('')        # remove the default suptitle
plt.ylabel('Salary (₹)')
plt.tight_layout()
plt.show()
```
