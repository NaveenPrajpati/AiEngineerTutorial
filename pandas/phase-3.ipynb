# Phase 3 — Data Analysis

Once data is clean, you analyse it by grouping, sorting, and summarising.

Topics:
- Sorting
- `groupby` — split / apply / combine
- Aggregation functions
- Pivot tables

```python
import pandas as pd
import numpy as np

# Sales dataset used throughout this phase
df = pd.DataFrame({
    'Name':       ['ram', 'shyam', 'karan', 'raju', 'priya', 'meena', 'arjun', 'sita'],
    'Department': ['IT',  'HR',    'IT',    'HR',   'IT',    'Finance','Finance','HR'],
    'City':       ['Delhi','Mumbai','Delhi','Pune', 'Mumbai','Delhi',  'Pune',  'Mumbai'],
    'salary':     [50000, 40000,  60000,  45000,  55000,   70000,   65000,   42000],
    'experience': [3,     5,      7,      2,      4,       9,       8,       3],
    'rating':     [4.2,   3.8,    4.5,    3.5,    4.0,     4.8,     4.6,     3.7]
})
print(df)
```

## 1. Sorting

`sort_values()` sorts by one or more columns.  
`ascending=False` for descending order.

```python
# Sort by single column
print("Sorted by salary (high to low):\n",
      df.sort_values('salary', ascending=False))

# Sort by multiple columns
print("\nSorted by Department then salary:\n",
      df.sort_values(['Department', 'salary'], ascending=[True, False]))
```

## 2. GroupBy — Split → Apply → Combine

`groupby` splits the DataFrame into groups, applies a function to each, then combines the results.  
Think of it like SQL's `GROUP BY`.

```python
# Average salary per department
print("Mean salary by Department:\n",
      df.groupby('Department')['salary'].mean())

# Count of employees per department
print("\nEmployee count by Department:\n",
      df.groupby('Department')['Name'].count())

# Max salary per department and city
print("\nMax salary by Dept & City:\n",
      df.groupby(['Department', 'City'])['salary'].max())
```

## 3. Aggregation with agg()

`agg()` lets you apply **multiple aggregation functions** at once, or different functions to different columns.

```python
# Multiple aggregations on one column
print("Salary stats by Department:\n",
      df.groupby('Department')['salary'].agg(['mean', 'min', 'max', 'count']))

# Different aggregations on different columns
print("\nCustom agg:\n",
      df.groupby('Department').agg(
          avg_salary  = ('salary',     'mean'),
          total_exp   = ('experience', 'sum'),
          avg_rating  = ('rating',     'mean'),
          headcount   = ('Name',       'count')
      ).round(2))
```

## 4. Value Counts

`value_counts()` counts how many times each unique value appears — great for categorical columns.

```python
print("Department distribution:\n", df['Department'].value_counts())
print("\nCity distribution:\n", df['City'].value_counts())

# Normalize to get percentages
print("\nDepartment % share:\n",
      df['Department'].value_counts(normalize=True).mul(100).round(1))
```

## 5. Pivot Table

A pivot table reshapes data — rows become one category, columns become another, and values are aggregated.

| Parameter | Meaning |
|---|---|
| `index` | Column to use as row labels |
| `columns` | Column to spread as column headers |
| `values` | Column whose values fill the table |
| `aggfunc` | How to aggregate (mean, sum, count, …) |

```python
# Average salary per Department × City
pivot = pd.pivot_table(
    df,
    index='Department',
    columns='City',
    values='salary',
    aggfunc='mean',
    fill_value=0
)
print("Pivot — avg salary by Dept × City:\n", pivot)

# Count of employees per Department × City
pivot_count = pd.pivot_table(
    df,
    index='Department',
    columns='City',
    values='Name',
    aggfunc='count',
    fill_value=0
)
print("\nPivot — headcount by Dept × City:\n", pivot_count)
```

## 6. Descriptive Statistics

`describe()` gives a quick summary. Pass `include='all'` to include string columns too.

```python
print("Numeric stats:\n", df.describe().round(2))
print("\nAll columns:\n", df.describe(include='all'))
```

## 7. Correlation

`corr()` shows how strongly numeric columns are related.  
Values range from **-1** (perfect inverse) to **+1** (perfect positive correlation).

```python
print("Correlation matrix:\n",
      df[['salary', 'experience', 'rating']].corr().round(2))
```
