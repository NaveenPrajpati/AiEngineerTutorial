# Phase 2 — Data Cleaning

Real-world data is messy. Phase 2 covers the most common cleaning tasks:
- Detecting and handling missing values (`NaN`)
- Removing duplicate rows
- Converting data types
- Renaming and reindexing

```python
import pandas as pd
import numpy as np
```

## 1. Missing Values (NaN)

`NaN` (Not a Number) represents missing data. Pandas uses `np.nan` internally.

| Method | What it does |
|---|---|
| `isnull()` | Returns True where value is NaN |
| `notnull()` | Returns True where value is NOT NaN |
| `dropna()` | Removes rows (or columns) with NaN |
| `fillna(value)` | Replaces NaN with a given value |

```python
df = pd.DataFrame({
    'Name':   ['ram', 'shyam', None,    'raju',  'priya'],
    'Age':    [25,    np.nan,  27,      30,      np.nan],
    'salary': [10000, 20000,   np.nan,  40000,   35000]
})
print(df)
```

```python
# Detect missing values
print("isnull():\n", df.isnull())
print("\nMissing count per column:\n", df.isnull().sum())
print("\nTotal missing:", df.isnull().sum().sum())
```

```python
# Drop rows that have ANY missing value
print("dropna():\n", df.dropna())

# Drop rows only if ALL values are missing
print("\ndropna(how='all'):\n", df.dropna(how='all'))

# Drop columns that have any missing value
print("\ndropna(axis=1):\n", df.dropna(axis=1))
```

```python
# Fill missing values with a constant
print("fillna(0):\n", df.fillna(0))

# Fill with the column mean — common for numeric columns
df_filled = df.copy()
df_filled['Age']    = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['salary'] = df_filled['salary'].fillna(df_filled['salary'].median())
df_filled['Name']   = df_filled['Name'].fillna('Unknown')
print("\nFilled with mean/median/Unknown:\n", df_filled)
```

## 2. Duplicate Rows

`duplicated()` marks duplicate rows as True. `drop_duplicates()` removes them.  
By default, the **first** occurrence is kept.

```python
df_dup = pd.DataFrame({
    'Name':   ['ram', 'shyam', 'ram',  'raju', 'shyam'],
    'Age':    [25,    30,      25,     28,     30],
    'salary': [10000, 20000,   10000,  35000,  20000]
})

print("Duplicate rows marked:\n", df_dup.duplicated())
print("\nNumber of duplicates:", df_dup.duplicated().sum())
print("\nAfter drop_duplicates():\n", df_dup.drop_duplicates())

# Check duplicates based on specific columns only
print("\nDuplicates by Name only:\n",
      df_dup.drop_duplicates(subset=['Name']))
```

## 3. Data Type Conversion

Pandas often reads numbers as strings from CSVs. Use `astype()` to fix types.  
`pd.to_numeric()` and `pd.to_datetime()` handle edge cases like bad values.

```python
df_types = pd.DataFrame({
    'Name':       ['ram', 'shyam', 'karan'],
    'Age':        ['25', '30', '27'],          # strings instead of ints
    'salary':     ['10000.5', '20000', '30000'],
    'join_date':  ['2020-01-15', '2019-06-20', '2021-03-10']
})

print("Before conversion:\n", df_types.dtypes)

df_types['Age']       = df_types['Age'].astype(int)
df_types['salary']    = df_types['salary'].astype(float)
df_types['join_date'] = pd.to_datetime(df_types['join_date'])

print("\nAfter conversion:\n", df_types.dtypes)
print("\nData:\n", df_types)
```

```python
# to_numeric with errors='coerce' — turns bad values into NaN instead of crashing
messy = pd.Series(['100', '200', 'abc', '400', None])
print("to_numeric coerce:\n", pd.to_numeric(messy, errors='coerce'))
```

## 4. String Cleaning

Access string methods via `.str` accessor — vectorized, no loops needed.

```python
df_str = pd.DataFrame({
    'Name':  ['  Ram ', 'SHYAM', 'karan  ', ' Raju'],
    'email': ['Ram@GMAIL.COM', 'shyam@yahoo.com', 'KARAN@gmail.com', 'raju@outlook.com']
})

df_str['Name']  = df_str['Name'].str.strip().str.title()   # remove spaces, title case
df_str['email'] = df_str['email'].str.lower()              # lowercase email
df_str['domain'] = df_str['email'].str.split('@').str[1]   # extract domain

print(df_str)
```

## 5. Resetting the Index

After filtering or dropping rows, the index keeps the original numbers (gaps).  
`reset_index(drop=True)` gives a clean 0-based index.

```python
df_sample = pd.DataFrame({
    'Name':   ['ram', 'shyam', 'karan', 'raju', 'priya'],
    'Age':    [25, 30, 27, 28, 22],
    'salary': [10000, 20000, 30000, 40000, 35000]
})

filtered = df_sample[df_sample['Age'] > 25]
print("Filtered (index has gaps):\n", filtered)

reset = filtered.reset_index(drop=True)
print("\nAfter reset_index:\n", reset)
```
