# Phase 4 — Combining DataFrames

Data often lives in multiple tables. Pandas provides three ways to combine them:

| Method | SQL equivalent | Use case |
|---|---|---|
| `pd.concat` | `UNION ALL` | Stack DataFrames vertically or horizontally |
| `pd.merge` | `JOIN` | Combine on matching key columns |
| `df.join` | `JOIN` on index | Combine on the index |

```python
import pandas as pd
import numpy as np
```

## 1. concat — Stacking DataFrames

`pd.concat` stacks DataFrames either **row-wise** (`axis=0`, default) or **column-wise** (`axis=1`).  
Use `ignore_index=True` to reset the index after stacking.

```python
df1 = pd.DataFrame({
    'Name':   ['ram', 'shyam'],
    'salary': [10000, 20000]
})

df2 = pd.DataFrame({
    'Name':   ['karan', 'raju'],
    'salary': [30000, 40000]
})

# Stack vertically (add more rows)
stacked = pd.concat([df1, df2], ignore_index=True)
print("Vertical concat:\n", stacked)

# Stack horizontally (add more columns)
df3 = pd.DataFrame({'Team': ['IT', 'HR', 'IT', 'HR']})
side_by_side = pd.concat([stacked, df3], axis=1)
print("\nHorizontal concat:\n", side_by_side)
```

## 2. merge — Joining on Key Columns

`pd.merge` is the most powerful combination tool. It works like SQL JOINs.

| how | Keeps rows from |
|---|---|
| `'inner'` (default) | only matching rows in both |
| `'left'` | all rows from left, matching from right |
| `'right'` | all rows from right, matching from left |
| `'outer'` | all rows from both, NaN where no match |

```python
employees = pd.DataFrame({
    'emp_id': [1, 2, 3, 4],
    'Name':   ['ram', 'shyam', 'karan', 'raju'],
    'dept_id': [10, 20, 10, 30]
})

departments = pd.DataFrame({
    'dept_id':   [10, 20, 40],
    'dept_name': ['IT', 'HR', 'Finance']
})

print("employees:\n", employees)
print("\ndepartments:\n", departments)
```

```python
# INNER JOIN — only employees with a matching department
print("inner merge:\n",
      pd.merge(employees, departments, on='dept_id', how='inner'))

# LEFT JOIN — all employees, NaN if department not found
print("\nleft merge:\n",
      pd.merge(employees, departments, on='dept_id', how='left'))

# OUTER JOIN — all rows from both tables
print("\nouter merge:\n",
      pd.merge(employees, departments, on='dept_id', how='outer'))
```

```python
# Merge on columns with different names using left_on / right_on
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3],
    'monthly_pay': [50000, 40000, 60000]
})

merged = pd.merge(employees, salaries,
                  left_on='emp_id', right_on='employee_id')
print(merged.drop(columns=['employee_id']))  # drop the redundant key
```

## 3. join — Combining on Index

`df.join` is a shorthand for merging on the **index**. Useful when DataFrames share a common index.

```python
info = pd.DataFrame({
    'Name': ['ram', 'shyam', 'karan']
}, index=[1, 2, 3])

scores = pd.DataFrame({
    'score': [88, 92, 75]
}, index=[1, 2, 3])

joined = info.join(scores)
print("Joined on index:\n", joined)
```

## 4. Handling Merge Conflicts — Suffixes

When both DataFrames have a column with the same name (besides the key), pandas adds suffixes `_x` and `_y`.  
You can customise these with the `suffixes` parameter.

```python
df_a = pd.DataFrame({'id': [1, 2], 'value': [100, 200]})
df_b = pd.DataFrame({'id': [1, 2], 'value': [300, 400]})

# Default suffixes
print("Default suffixes:\n",
      pd.merge(df_a, df_b, on='id'))

# Custom suffixes
print("\nCustom suffixes:\n",
      pd.merge(df_a, df_b, on='id', suffixes=('_before', '_after')))
```
