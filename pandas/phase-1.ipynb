# Phase 1 — Pandas Basics

Pandas is built on top of NumPy and provides two key data structures:
- **Series** — a 1D labeled array (like a single column)
- **DataFrame** — a 2D table with labeled rows and columns (like a spreadsheet)

```
import pandas as pd
```

````python
df=pd.DataFrame([1,2,4],columns=['col_name'])
print(df)
print(type(df))
````

## 1. Creating a DataFrame

You can create a DataFrame from a list, dictionary, NumPy array, or CSV file.

````python
# Creating from a dictionary — most common approach
data = {
    'Name':   ['ram', 'shyam', 'karan', 'raju'],
    'Age':    [12, 25, 27, 30],
    'salary': [10000, 20000, 30000, 40000]
}
````

````python
df=pd.DataFrame(data)
print(df)
````

## 2. Basic DataFrame Properties

These attributes give you a quick overview of a DataFrame's structure and contents.

````python
print("head(2):\n", df.head(2))       # first 2 rows
print("\ntail(2):\n", df.tail(2))      # last 2 rows
print("\nshape   :", df.shape)         # (rows, columns)
print("columns :", df.columns.tolist())
print("index   :", df.index.tolist())
print("dtypes  :\n", df.dtypes)       # data type of each column
````

````python
# Without inplace — returns a new DataFrame, original unchanged
renamed = df.rename(columns={'salary': 'monthly_salary'})
print("renamed (new df):\n", renamed)

# With inplace=True — modifies df directly, returns None
df.rename(columns={'salary': 'monthly_salary'}, inplace=True)
print("\ndf after inplace rename:\n", df)
````

````python
print(df.info())
````

````python
print(df.describe())
````

## 3. Save & Load Data from CSV

`to_csv` exports the DataFrame. `read_csv` reads it back.  
`index=False` prevents pandas from writing the row numbers as an extra column.

````python
df.to_csv('test_data.csv',index=False)
````

````python
loaded=pd.read_csv('test_data.csv')
print(loaded)
````

## 4. Row & Column Selection

Two main ways to select data:
- **`loc`** — label-based (uses column names, **includes** the stop index)
- **`iloc`** — position-based (uses integers, **excludes** the stop index like Python slices)

````python
# Select a single column
print("Single column:\n", df[['Name']])

# loc — label based, stop index INCLUDED
print("\nloc[0:2]:\n", df.loc[0:2])

# iloc — position based, stop index EXCLUDED (like Python lists)
print("\niloc[0:2]:\n", df.iloc[0:2])

# Single row by position
print("\niloc[0] (single row):\n", df.iloc[0])

# Row by condition
print("\nWhere Name == 'ram':\n", df.loc[df['Name'] == 'ram'])
````

## 5. Filtering a DataFrame

Use boolean conditions inside `[]` to filter rows.  
Combine conditions with `&` (and) / `|` (or) — always wrap each condition in parentheses.

````python
# Single condition
filtered = df[df['Age'] > 25]
print("Age > 25:\n", filtered)

# Multiple conditions — use & (and), | (or)
print("\nAge >= 23 AND monthly_salary > 30000:\n",
      df[(df['Age'] >= 23) & (df['monthly_salary'] > 30000)])

# where — keeps shape, replaces non-matching with NaN or a fill value
print("\ndf.where(Age >= 25) — non-matching rows become NaN:\n", df.where(df['Age'] >= 25))

print("\ndf.where(Age >= 25, other='Not Eligible'):\n",
      df.where(df['Age'] >= 25, other='Not Eligible'))
````

## 6. Adding & Removing Rows / Columns

````python
# Restart with a clean df so this cell always works independently
df = pd.DataFrame({
    'Name':           ['ram', 'shyam', 'karan', 'raju'],
    'Age':            [12, 25, 27, 30],
    'monthly_salary': [10000, 20000, 30000, 40000]
})

# Add a new column
df['Team'] = ['CEO', 'HR', 'CTO', 'DA']
print("After adding Team:\n", df)

# Derived column from existing columns
df['Bonus'] = df['monthly_salary'] * 0.2
print("\nAfter adding Bonus:\n", df)

# Add a new row using loc
df.loc[len(df)] = ['rahul', 20, 21000, 'EMPL', 4200]
print("\nAfter adding a row:\n", df)

# Update a single cell value
df.loc[0, 'Age'] = 19
print("\nAfter updating ram's Age to 19:\n", df)

# Drop a column
df_no_bonus = df.drop(columns=['Bonus'])
print("\nAfter dropping Bonus column:\n", df_no_bonus)

# Drop a row by index
df_no_first = df.drop(index=0)
print("\nAfter dropping row 0:\n", df_no_first)
````
